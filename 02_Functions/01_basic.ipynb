```python
# have a discount 

def applydiscount(bill):
    discount=bill*0.35
    final_bill=bill-discount
    print("Discount is :: ",discount)
    print("Final BIll After Discount is :: ",final_bill)

applydiscount(50000)
applydiscount(90000)
```

```python
# factorial
def factorial():
    num=int(input("Enter number :: "))
    fact=1
    for i in range(1,num+1):
        fact=fact*i
        i=i+1

    print("Factorial is :: ",fact)

factorial()
factorial()
```
